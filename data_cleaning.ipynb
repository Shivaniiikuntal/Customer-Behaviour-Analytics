```python
import pandas as pd
import numpy as np
```

```python
data = pd.read_csv("Dataset/marketing_campaign.csv", sep="\t")
data.head()
```

```python
data.shape
```

```python
data.describe()
```

```python
data.isnull().sum()
```

```python
data["Income"].fillna(data["Income"].mean(), inplace = True)
```

```python
data["Dt_Customer"] = pd.to_datetime(data["Dt_Customer"], format="%d-%m-%Y")
```

```python
data["Age"] = 2025 - data["Year_Birth"]
data.head()
```

```python
data["Customer_for"] = (pd.to_datetime("today") - data["Dt_Customer"]).dt.days / 365
data["Customer_for"] = np.round(data["Customer_for"], 2)
data["Customer_for"].head()
```

```python
data["Marital_Status"].value_counts()
```

```python
data['Marital_Status'] = data['Marital_Status'].replace({
    'Together': 'Partnered',
    'Married': 'Partnered',
    'Single': 'Single',
    'Alone': 'Single',
    'YOLO': 'Single',
    'Absurd': 'Single'
})
```

```python
data["Children"] = data["Kidhome"] + data["Teenhome"]
data["Family_Size"] = data["Children"] + 1
data.loc[data["Marital_Status"] == "Partenered", "Family_Size"]+=1
```

```python
data['Total_Spending'] = np.sum(data[['MntWines','MntFruits','MntMeatProducts',
                                  'MntFishProducts','MntSweetProducts','MntGoldProds']].values, axis=1)

data['Total_Accepted_Cmp'] = np.sum(data[['AcceptedCmp1','AcceptedCmp2','AcceptedCmp3',
                               'AcceptedCmp4','AcceptedCmp5','Response']].values, axis=1)
```

```python
average_income = np.mean(data["Income"])
median_income = np.median(data["Income"])
std_income = np.std(data["Income"])
var_income = np.var(data["Income"])
q1, q3 = np.percentile(data['Income'], [25, 75])
print(f"Average Income: {average_income}\nMedian Income: {median_income}\nStandard Deviation: {std_income}\nVariance: {var_income}\nQ1: {q1}\nQ3: {q3}")
```

```python
min_age, max_age = np.min(data['Age']), np.max(data['Age'])
total_spending = np.sum(data['Total_Spending'])
print(f"Minimum Age: {min_age}\nMaximum Age: {max_age}\nTotal Spending: {total_spending}")
```

```python
q1_age = data['Age'].quantile(0.25)
q3_age = data['Age'].quantile(0.75)
IQR_age = q3_age - q1_age
lower_bound = q1_age - 1.5 * IQR_age
upper_bound = q3_age + 1.5 * IQR_age
outliers_age = data[(data["Age"] < lower_bound) | (data["Age"] > upper_bound)]
outliers_age
```

```python
median_age = data['Age'].median()
data.loc[(data['Age'] < lower_bound) | (data['Age'] > upper_bound), 'Age'] = median_age
```

```python
for col in ["Income", "Total_Spending"]:
    median_val = data[col][data[col] > 0].median()
    data.loc[data[col] < 0, col] = median_val
```

```python
q1_spend = data['Total_Spending'].quantile(0.25)
q3_spend = data['Total_Spending'].quantile(0.75)
IQR_spend = q3_spend - q1_spend
lower_bound = q1_spend - 1.5 * IQR_spend
upper_bound = q3_spend + 1.5 * IQR_spend
outliers_spend = data[(data['Total_Spending'] < lower_bound) | (data['Total_Spending'] > upper_bound)]
outliers_spend
```

```python
data['Total_Spending_Log'] = np.log1p(data['Total_Spending'])
data["Income_Log"] = np.log1p(data['Income'])
```

```python
q1_spend = data['Total_Spending_Log'].quantile(0.25)
q3_spend = data['Total_Spending_Log'].quantile(0.75)
IQR_spend = q3_spend - q1_spend
lower_bound = q1_spend - 1.5 * IQR_spend
upper_bound = q3_spend + 1.5 * IQR_spend
outliers_spend = data[(data['Total_Spending_Log'] < lower_bound) | (data['Total_Spending_Log'] > upper_bound)]
outliers_spend
```

```python
from sklearn.preprocessing import StandardScaler
standard_scaler = StandardScaler()

purchase_cols = ["NumDealsPurchases", "NumWebPurchases",
                 "NumCatalogPurchases", "NumStorePurchases", "NumWebVisitsMonth"]

data[purchase_cols] = standard_scaler.fit_transform(data[purchase_cols])
```

```python
from sklearn.preprocessing import MinMaxScaler
min_max_scaler = MinMaxScaler()
data["Recency_Scaled"] = min_max_scaler.fit_transform(data[["Recency"]])
```
